```python

my_list = [ 'ram',100, 'hari', 6,'shyam',-1,None, "Null", 999,799]


num = []


for i in my_list:
    if type(i) == str:
            if i == "Null":
                continue
            num.append(i)
    elif type(i) == int:
        if i>100 or i<0:
            continue
        num.append(i)
        


print( num)
```
